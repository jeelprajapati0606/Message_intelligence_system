# Part A : Probability & Conceptual Foundation (Theory)

## Q1. What is Conditional Probability?

####  Conditional probability is the probability of an event occurring when another event has already occurred. It helps us predict outcomes based on given conditions.

**Formula** :  P(A|B) = P(A and B) / P(B)

## Q2. Explain Bayes' Theorem and its Importance.

####  Bayes' Theorem is used to calculate the probability of an event based on prior knowledge of related conditions. 

####  It is important in classification because it helps predict whether a message is spam or legitimate.

**Formula**: P(A|B) = [P(B|A) × P(A)] / P(B)

## 3. What assumptions does the Naive Bayes Classifier make?

The **Naive Bayes Classifier** is a machine learning algorithm based on **Bayes' Theorem**. It is used to predict the class of data.

It makes the following assumptions:

1. **Feature Independence:** It assumes that all features are independent of each other.
2. **Equal Contribution:** Each feature contributes information to predict the class based on its probability.
3. **Probability-Based Prediction:** It calculates the probability of each class and selects the class with the highest probability.

**Example:** To predict whether an email is spam, Naive Bayes checks words like "free", "offer", and "win". It assumes that these words are independent of each other.

## 4. Explain the working principle of K-Nearest Neighbors (KNN) and Support Vector Machine (SVM).

### A. K-Nearest Neighbors (KNN)

**KNN** is a supervised machine learning algorithm that classifies data based on its nearest neighbors.

**Working Principle:**

1. Choose the value of **K**, which represents the number of nearest neighbors.
2. Calculate the distance between the new data point and the existing data points.
3. Find the **K nearest neighbors**.
4. Count the classes of these neighbors.
5. Assign the class that gets the majority of votes.

**Example:** If 3 out of the 5 nearest neighbors are cats and 2 are dogs, KNN predicts that the new animal is a **cat**.

### B. Support Vector Machine (SVM)

**SVM** is a supervised machine learning algorithm that separates data into different classes using a decision boundary.

**Working Principle:**

1. SVM takes the training data with different classes.
2. It finds a **decision boundary (hyperplane)** that separates the classes.
3. It tries to maximize the margin between the boundary and the nearest data points.
4. The nearest data points are called **support vectors**.
5. The new data point is classified based on which side of the boundary it falls on.

**Example:** SVM separates emails into two classes: spam and not spam, by finding the best boundary between them.

## 5. Compare distance-based, probabilistic, and margin-based classifiers.

| Basis | Distance-Based | Probabilistic | Margin-Based |
|---|---|---|---|
| **Meaning** | Uses distance between data points. | Uses probability to predict a class. | Uses a boundary to separate classes. |
| **Working Principle** | Finds the nearest data points. | Calculates the probability of each class. | Finds the best boundary with the maximum margin. |
| **Main Algorithm** | KNN | Naive Bayes | SVM |
| **Prediction** | Based on nearest neighbors. | Based on the highest probability. | Based on the decision boundary. |
| **Advantages** | Simple and easy to understand. | Fast and works well with high-dimensional data. | Works well with clear class boundaries. |
| **Disadvantages** | Can be slow on large datasets and sensitive to feature scales. | Assumptions may not always be true. | Can be slow on large datasets and needs parameter tuning. |
| **Example** | Classifying a fruit using similar fruits. | Classifying an email as spam or not spam. | Separating cats and dogs using their features. |

**Conclusion:** Distance-based classifiers use distance, probabilistic classifiers use probability, and margin-based classifiers use a decision boundary to classify data.

# Part B: Dataset Understanding & Preparation

## Q6. Identify Input Features and Target Variable

```python
import pandas as pd

df = pd.read_csv("Message_Intelligence_Dataset_5200 - Message_Intelligence_Dataset_5200.csv.csv")

# Input Features
X = df.drop("spam_label", axis=1)

# Target Variable
y = df["spam_label"]

print("Features:")
print(X.columns)

print("Target:")
print(y.name)
```

## Q7. Perform Basic Data Preprocessing

```python
# String columns remove
X = X.drop(["message_text", "timestamp"], axis=1)

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

print("Data Preprocessed Successfully")
```

## Q8. Split Dataset into Training and Testing Sets

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled,
    y,
    test_size=0.2,
    random_state=42
)

print("Training Shape:", X_train.shape)
print("Testing Shape:", X_test.shape)
```

# Part C: Baseline Model - K Nearest Neighbors (KNN)

## Q9. Implement KNN Classifier

```python
# missing value fill mean

from sklearn.impute import SimpleImputer

imputer = SimpleImputer(strategy='mean')

X = imputer.fit_transform(X)

# scaling

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

# split

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled,
    y,
    test_size=0.2,
    random_state=42
)

# KNN

from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier(n_neighbors=5)

knn.fit(X_train, y_train)

y_pred_knn = knn.predict(X_test)

print("KNN Model Created")
```

## Q10. Experiment with Different Values of K

```python
from sklearn.metrics import accuracy_score

for k in [1,3,5,7,9]:
    
    model = KNeighborsClassifier(n_neighbors=k)
    
    model.fit(X_train, y_train)
    
    pred = model.predict(X_test)
    
    accuracy = accuracy_score(y_test, pred)
    
    print("K =", k, "Accuracy =", accuracy)
```

## Q11. Analyze Distance Metrics

```python
for metric in ["euclidean","manhattan"]:
    
    model = KNeighborsClassifier(
        n_neighbors=5,
        metric=metric
    )
    
    model.fit(X_train,y_train)
    
    pred = model.predict(X_test)
    
    acc = accuracy_score(y_test,pred)
    
    print(metric,"Accuracy =",acc)
```
